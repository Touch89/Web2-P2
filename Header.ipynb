Login para el token

```python
import requests

res = requests.post("http://localhost:8000/login", data={
    "username": "eduardo",
    "password": "contraseña"
})
token = res.json()["access_token"]
print("Token:", token)
```

Llamar ruta con el token en el header

```python
res = requests.get("http://localhost:8000/users/me", headers={
    "Authorization": f"Bearer {token}"
})
print(res.json())
```
